## **<span style = "color:red">Question</span>: How do you optimize a slow Power BI report?**

**Use a structured approach:**

1. Identify **slow visuals** using **Performance Analyzer**.
2. Inspect the **semantic model** and relationships.
3. Prefer a **star schema** over unnecessarily complex relationships.
4. Remove **unused columns** and reduce **unnecessary cardinality**.
5. **Optimize DAX** and avoid expensive iterator operations when simpler expressions work.
6. Use **Import, DirectQuery**, or **Direct Lake based** on data volume, freshness, and operational requirements.
7. **Optimize** the underlying **SQL queries** or **Lakehouse tables**.
8. **Check capacity** and refresh performance.

I approach Power BI performance from both the **semantic-model** layer and the **data-source** layer. A **slow visual** may be caused by **inefficient DAX**, a poorly designed model, expensive source queries, or **capacity limitations**, so I identify the bottleneck before changing the architecture.

    <tr><th>Mode</th><th>How it works</th><th>Best use cases</th></tr>
    <tr><td>Import</td><td>Data is loaded into the Power BI model</td><td>Fast interactive reporting</td></tr>
    <tr><td>DirectQuery</td><td>Queries are sent to the underlying source</td><td>When source-level querying or freshness requirements justify it</td></tr>
    <tr><td>Direct Lake</td><td>Reads Delta tables in OneLake through the Fabric semantic model</td><td>Large Fabric-based analytical solutions</td></tr>

</table>

Mention that Direct Lake is not simply identical to Import or DirectQuery. Its behavior, fallback considerations, and performance depend on the Fabric semantic model and its underlying data.
