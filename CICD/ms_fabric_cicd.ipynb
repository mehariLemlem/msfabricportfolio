# **Microsoft Fabric CI/CD**

- For a Senior Data Engineer interview, CI/CD in Microsoft Fabric is about much more than moving a notebook from one workspace to another. You need to <span style = "color:green">**demonstrate**</span> how you automate deployments, <span style = "color:green">**manage**</span> source control, <span style = "color:green">**promote**</span> changes safely across environments, <span style = "color:green">**protect**</span> production data, <span style = "color:green">**validate**</span> releases, and <span style = "color:green">**recover**</span> from failed deployments.

## **CI/CD Architecture**

1. <span style = "color:orange">**Development workspace:**</span> Engineer develops notebooks, pipelines, Lakehouses, and semantic models.
    - Commit and push changes
2. <span style = "color:orange">**Azure DevOps Git repository:**</span> Feature branches, pull requests, code reviews, version history.
    - Automated validation and approval
3. <span style = "color:orange">**CI validation:**</span> Python tests, SQL checks, configuration checks, security and quality gates.
    - Release artifact and deployment
        - <span style = "color:green">**Test**</span> (Integration tests) → <span style = "color:green">**UAT**</span> (Business approval) → <span style = "color:green">**Production**</span> (Controlled release)

<span style = "color:red">**Important distinction:**</span> Fabric Git integration, Fabric deployment pipelines, and Azure DevOps pipelines are related but different capabilities.
- <span style = "color:orange">**Git integration:**</span> tracks supported Fabric item definitions and enables collaboration with a repository.
- <span style = "color:orange">**Fabric deployment pipelines:**</span> promote supported items between Fabric workspaces, such as Development, Test, and Production.
- <span style = "color:orange">**Azure DevOps CI/CD pipelines:**</span> automate validation, orchestration, testing, approvals, and deployment using supported APIs, tooling, and scripts.

A strong senior-level answer explains how these capabilities work together instead of treating them as interchangeable.

## **Common CI/CD Interview Questions**

### **What is CI/CD, and why is it important in Microsoft Fabric?**

CI/CD stands for Continuous Integration and Continuous Delivery or Deployment.
- <span style = "color:orange">**Continuous Integration (CI):**</span> Developers frequently commit changes to source control, where the team reviews and validates them.
- <span style = "color:orange">**Continuous Delivery:**</span> Validated changes are prepared for release, with production deployment subject to approval.
- <span style = "color:orange">**Continuous Deployment:**</span> Changes that pass all required checks are deployed automatically without a manual production approval.
In Microsoft Fabric, CI/CD helps us manage notebooks, pipelines, Lakehouses, semantic models, and other supported artifacts consistently across environments.
Real-world example: Suppose I modify a PySpark notebook that transforms healthcare claims. Before releasing it, I validate the code, run unit tests against sample data, verify the output schema, deploy to Test, and compare record counts and business rules. After UAT approval, I promote the release to Production.
This reduces deployment errors, improves traceability, and makes releases repeatable.

### **How do you manage Git branches for a Fabric project?**

I would use a branching strategy that supports parallel development while protecting the release branch.

For example:

- main: approved, production-ready code.
- feature/claims-validation: a developer's new functionality.
- feature/member-dedup: a separate developer's changes.
- release/2026.10: an optional stabilization branch for a coordinated release.
- hotfix/claims-null-handling: an urgent production correction.

The developer creates a feature branch, makes changes, tests locally or in Development, and opens a pull request. Another engineer reviews the code, and automated validation must pass before merging.
I would enable branch policies, required reviewers, and build validation where appropriate.

**Why it matters:** Without branch protection, one developer can accidentally introduce untested notebook changes or overwrite another developer's work.

### **How would you implement automated testing in a Fabric CI/CD pipeline?**

I would implement several layers of testing.

1. Unit tests
    - Validate individual transformation functions, null handling, deduplication, and business logic.
2. Integration tests
    - Verify that ingestion, transformation, storage, and downstream dependencies work together.
3. Data quality tests
    - Validate required columns, uniqueness, referential integrity, record counts, accepted values, and freshness.
4. Performance and security tests
    - Check execution time, resource use, access controls, and exposure of sensitive data.


<table>
<td>

For example, suppose a notebook produces a Silver claims table. I would test that:
- ClaimID is not null.
- Each claim meets the expected uniqueness rule.
- PaidAmount is nonnegative unless reversals explicitly permit negative values.
- Every MemberID references a valid member.
- The transformed row count reconciles with the expected source population after documented filtering.

</td>
</table>

```python
# A simple PySpark validation could be:
#=====================================

from pyspark.sql import functions as F

claims = spark.table("silver_claims")

null_claim_ids = claims.filter(
    F.col("ClaimID").isNull()
).count()

duplicate_claim_ids = (
    claims.groupBy("ClaimID")
    .count()
    .filter(F.col("count") > 1)
    .count()
)

if null_claim_ids > 0 or duplicate_claim_ids > 0:
    raise ValueError(
        f"Data quality failure: "
        f"null IDs={null_claim_ids}, "
        f"duplicate IDs={duplicate_claim_ids}"
    )
```

The CI/CD process should fail the relevant release gate when a required test fails. In practice, I would separate tests that can run outside Fabric from tests that require an actual Fabric workspace or Spark session.

### **How do you handle database schema changes during deployment?**

I treat schema changes as controlled data migrations, not just code deployments.
Suppose I add a new column, **ClaimCategory**, to a Gold claims table consumed by a Direct Lake semantic model.

<table>
<tr>
<td>
I would:

1. Determine the impact on downstream notebooks, reports, measures, and consumers.
2. Prefer backward-compatible changes where possible.
3. Deploy the updated transformation and schema in the correct dependency order.
4. Backfill historical data if required.
5. Validate the schema, data quality, and downstream measures.
6. Remove obsolete columns only after dependent consumers have migrated.

</td>
</tr>
</table>

For breaking changes, I may use an expand-and-contract approach: introduce the new field, migrate consumers, then remove the old field in a later release.

<span style = "color:red">**Important:**</span> Deploying a notebook definition does not automatically migrate all underlying Delta table data or guarantee that every downstream consumer is compatible. Data migration, table evolution, and semantic model validation need explicit planning.

### **How do you manage credentials and security in CI/CD?**

I follow **least privilege** and separate identities and permissions by environment.

- Use a <span style = "color:orange">**managed identity**</span> or <span style = "color:orange">**service principal**</span> where supported by the specific Fabric operation.
- Store secrets in Azure Key Vault or an approved secrets-management system.
- Use separate development and production permissions.
- Avoid embedding passwords, access tokens, or connection secrets in notebooks and YAML files.
- Audit deployment activity and protect production access.
- Validate RLS, CLS, and other relevant data-access controls after deployment.

For example, a deployment identity may have permission to publish approved artifacts without having unrestricted access to healthcare claims data.

<span style = "color:orange">**Senior-level distinction:**</span> Deployment authorization and runtime data access are separate concerns. A service principal that can deploy an artifact does not automatically have every permission needed to configure OAuth connections, refresh a semantic model, or read a protected data source. I verify the supported authentication method for each operation.

    <tr><th>Feature</th><th>Managed Identity</th><th>Service principal</th></tr>

### **How do you handle a failed production deployment?**

I use a controlled recovery process instead of immediately rerunning the entire pipeline.

<table>
<td>

1. **Stop further promotion**. Prevent the failed release from reaching additional consumers or environments.
2. **Identify the failure point**. Review Azure DevOps logs, Fabric deployment results, pipeline activity, Spark logs, and dependency errors.
3. **Assess the impact**. Determine whether the failure affected only artifact deployment, pipeline execution, data integrity, or report availability.
4. **Restore service**. Redeploy a known-good artifact version or apply an approved corrective change.
5. **Validate recovery**. Run smoke tests, reconcile critical data, verify permissions, and confirm downstream reports work.
6. **Prevent recurrence**. Add the missing automated test, dependency check, or release control.

</td>
</table>


A key distinction is that rolling back code does not automatically roll back data. If a faulty notebook overwrote production data, I would also need a data recovery strategy appropriate to the storage system and its available recovery capabilities.

### **<span style = "color:red">Scenario:</span> A developer changes a notebook, and the Production pipeline fails. The notebook passed Development testing but failed in Production. How do you investigate?**

<table>
<td>

I would inspect the Spark exception and pipeline activity logs, reproduce the issue using representative data, and determine whether the problem is <span style = "color:orange">**configuration-related, data-related, or code-related**</span>.
I would restore the previous working release if business operations were affected, then fix the issue in <span style = "color:orange">**source control**</span> and add a <span style = "color:orange">**regression test**</span> before redeploying.

### **Rapid-fire questions to practice**

1. Can you explain CI versus CD?CI validates integrated code changes;
    - CD prepares or deploys validated changes through controlled release stages.
2. Does Git integration deploy data?
    - Git integration tracks supported item definitions and changes, not a general backup or deployment mechanism for Lakehouse table contents.
3. What is a pull request?
    - A controlled review of proposed changes before merging them into a shared branch.
4. What is a deployment rule?
    - A stage-specific setting that overrides supported configuration, such as a semantic model data source or a notebook's default Lakehouse, where applicable.
5. What is a smoke test?
    - A quick post-deployment check confirming that critical components work.
6. How do you roll back?
    - Restore a known-good artifact version and separately recover data if necessary.
7. Should secrets be committed to Git?
    - No. Use an approved secrets-management solution and supported identity-based authentication.
8. What is infrastructure as code?
    - Defining and managing provisionable infrastructure through version-controlled configuration rather than relying entirely on manual setup.
9. What is idempotency?
    - Rerunning the same operation with the same inputs produces the same intended state without unintended duplicate effects.
10. How do you handle schema drift?
    - Detect schema changes, validate compatibility, quarantine or reject unexpected changes where necessary, and update downstream contracts deliberately.
11. What is a release artifact?
    - The versioned code or deployable definition associated with a specific release, along with the metadata needed to reproduce it.
12. Why separate workspaces?
    - To isolate permissions, configurations, testing, and production workloads.

I approach CI/CD in Microsoft Fabric as an end-to-end release process, not just a deployment activity.

I use Git integration with Azure DevOps to manage version control, feature branches, pull requests, and peer reviews. I then use automated validation to check code quality, transformation logic, schema compatibility, and data quality.

For deployment, I use Fabric deployment pipelines or supported API-driven automation to promote artifacts through Development, Test, and Production. I manage environment-specific configurations separately, secure credentials using approved identity and secrets-management mechanisms, and validate dependencies and target-stage bindings.

My goal is to make releases repeatable, secure, observable, and recoverable while minimizing production risk.
