### **Workspace monitoring**

<table>
<td>

**Enable Workspace Monitoring:**
- <span style = "color:green">Workspace ⇒ Workspace settings ⇒ Monitoring ⇒ +Eventhouse ⇒ switch ON the "**Log workspace activity**" toggle.</span>
- With logging on, workspace activity data is collected and stored in the read-only monitoring KQL database within the monitoring Eventhouse. Users can query the database for performance and diagnostic information.

</td>
</table>

- When you **enable workspace monitoring**, Fabric creates an **Eventhouse database** in your workspace that **continuously collects diagnostic logs** and **metrics** from your Fabric items.
- Workspace monitoring stores raw log data that you can query with **KQL** or **SQL**.
- A workspace can use either **workspace monitoring** or **Log Analytics**, not both at once. If Log Analytics is already configured, delete that configuration and wait a few hours first.

- It needs a setup in each workspace. It isn’t tenant-wide; each workspace admin turns it on.
- Menu names change. Microsoft updates the Fabric UI often, so if a label differs from the above, follow the current Learn article “Enable monitoring in your workspace.”

### **Activator**

- The **monitoring hub** shows you what happened. The **Schedule failures** page emails you when it happens. But neither takes action for you. **Activator** in Microsoft Fabric closes that gap — when a Fabric job event occurs, Activator can automatically notify the right people in the right channel, or kick off a follow-up job without waiting for a human response.
- **When the condition is met, Activator can take these actions:**
    - **Email:** Sends a message with context about the event, including a link to the monitoring hub.
    - **Teams:** Posts a message to an individual, group chat, or channel. This is often more useful than email for on-call teams who need immediate visibility.
    - **Run a Fabric activity:** Triggers a pipeline, notebook, dataflow, spark job, or function.
- **Activator works best for automated coordination:** chaining jobs together, running cleanup actions, or getting the right notification to the right channel — not as a substitute for investigating and fixing the root cause.



<table>
<td>

**The quickest way to connect Activator to a job event is through Real-Time hub, using the Set alert wizard:**

1. In Microsoft Fabric, **select Real-Time** in the left navigation to open Real-Time hub.
2. Select **Fabric events**, then select Job events.
3. Select **Set alert**.
4. In the **Monitor** section, select **Select source events** and complete the wizard — choose the event type (for example, <span style = "color:orange">**Microsoft.Fabric.JobEvents.ItemJobFailed**</span>), then select the workspace and the specific item to monitor.
5. In the **Condition** section, choose when to act — on each matching event, or only when a specific field value is met.
6. In the **Action** section, choose what Activator should do.


</td>
</table>

## **Secure Data Access in Microsoft Fabric**

&#8505;&#65039; <span style = "color:brown">**Layered security model / Multi-layer security model: (Three levels of access evaluation)**</span>

1. <span style = "color:blue">**Microsoft Entra ID authentication:**</span> checks whether the user can authenticate with Microsoft Entra ID.
2. <span style = "color:blue">**Fabric access:**</span> checks whether the user can access Fabric.
3. <span style = "color:blue">**Data security:**</span> checks whether the user can perform the requested action on a table or file.
    - <span style = "color:red">**Workspace roles**</span> (apply to all items in a workspace): ([Read more](https://learn.microsoft.com/en-us/fabric/fundamentals/roles-workspaces))
        - <span style = "color:orange">**Admin:**</span> Can create, view, modify, share, and <span style = "color:green">**manage**</span> all content and data in the workspace, and manage permissions.
        - <span style = "color:orange">**Member:**</span> Can create, view, modify, and <span style = "color:green">**share**</span> all content and data in the workspace.
        - <span style = "color:orange">**Contributor:**</span> Can <span style = "color:green">**create**</span>, view, and <span style = "color:green">**modify**</span> all content and data in the workspace.
        - <span style = "color:orange">**Viewer:**</span> Can view all content in the workspace, but can't modify it.
            <table>
            <td>

            <span style = "color:brown">**Add users to workspace role:**</span> In the workspace, select **Manage access** ⇒ Enter the user's **name or email address** ⇒ Select the workspace role to **assign** and select **Add**.

            </td>
            </table>
    - <span style = "color:red">**Item permissions**</span> (apply to a single item, like a specific lakehouse or warehouse) ([Read more](https://learn.microsoft.com/en-us/fabric/security/permission-model#item-permissions))

        - **[Lakehouse](https://learn.microsoft.com/en-us/fabric/data-engineering/lakehouse-sharing):**
            - **Read all SQL endpoint data:** Read data from the SQL analytics endpoint using T-SQL
            - **Read all Apache Spark and subscribe to events:** Read lakehouse data through Apache Spark and OneLake APIs
            - **Build reports on the default semantic model:** Create Power BI reports on the default semantic model
        - **[Semantic Model](https://learn.microsoft.com/en-us/power-bi/connect-data/service-datasets-permissions)**

        - **[Warehouse](https://learn.microsoft.com/en-us/fabric/data-warehouse/share-warehouse-manage-permissions)**

        - **[Data Factory](https://learn.microsoft.com/en-us/fabric/data-factory/data-factory-overview)**

        - **[SQL Database](https://learn.microsoft.com/en-us/fabric/database/sql/security-overview)**

        - **[Data Science](https://learn.microsoft.com/en-us/fabric/data-science/models-experiments-rbac)**

        - **[Real-Time Intelligence](https://learn.microsoft.com/en-us/kusto/management/security-roles)**

        - **[Mirrored Database](https://learn.microsoft.com/en-us/fabric/mirroring/share-and-manage-permissions)**
        
            <table>
            <td>

            <span style = "color:brown">**Share an item and configure its permissions:**</span> select the **ellipsis (...)** next to the item in the workspace ⇒ select **Manage permissions** ⇒ Select **Permissions** ⇒ select **Grant access**
    

            </td>
            </table>

    - <span style = "color:red">**Compute or granular permissions**</span> (apply within a specific engine, like the SQL analytics endpoint or semantic model)
        - **T-SQL permissions** (compute permissions for the SQL analytics endpoint)
            - **[GRANT](https://learn.microsoft.com/en-us/sql/t-sql/statements/grant-transact-sql)**
            - **[DENY](https://learn.microsoft.com/en-us/sql/t-sql/statements/deny-transact-sql)**
            - **[REVOKE](https://learn.microsoft.com/en-us/sql/t-sql/statements/revoke-database-permissions-transact-sql)**
            - **[Row-level security](https://learn.microsoft.com/en-us/fabric/data-warehouse/row-level-security)**
            - **[Column-level security](https://learn.microsoft.com/en-us/fabric/data-warehouse/column-level-security)**
            - **[Dynamic Data Masking (DDM)](https://learn.microsoft.com/en-us/fabric/data-warehouse/dynamic-data-masking)**

    - <span style = "color:red">**OneLake security**</span> (adds control within an item. These restrictions apply consistently across all compute engines — Spark, SQL, and OneLake APIs. Use OneLake security when a user needs access to a Fabric item but should only see specific data within it.)
        - Users with **Admin**, **Member**, or **Contributor** workspace roles already have full read and write access to all OneLake data — OneLake security roles don't restrict that access.
        - Use OneLake security roles to grant specific data access to users in the **Viewer workspace role** or **with only Read item permission**, who have no OneLake data access by default.
        - Only workspace **Admin** and **Member** roles can create or modify OneLake security roles.

        <table>
        <td>

        &#8505;&#65039; <span style = "color:blue">**Each OneLake security role has four components:**</span>

        - **Data:** The tables or folders that users can access.
        - **Permission:** The level of access: Read (view data) or ReadWrite (view and edit data in specific tables or folders without granting a workspace write role).
        - **Members:** The users or groups assigned to the role.
        - **Constraints:** Optional row or column filters applied to the data in the role.

        </td>
        </table>

<table>
<td>

**&#8505;&#65039; <span style = "color:red">Note:</span>**

- **Viewers** can see Fabric items listed in the workspace, but have **no access** to the underlying data stored in **OneLake by default**. Use OneLake security roles to grant Viewers access to specific tables or folders.
- Every new lakehouse automatically includes a **DefaultReader** role. When you **share** a lakehouse with the **Read all Apache Spark and subscribe to events** permission, the recipient is added to **DefaultReader** and gets read access to all data. When you **restrict a user to a custom role**, also **remove them from DefaultReader** — otherwise they still have full read access through the default role.



</td>
</table>

<table>
<td>

**&#8505;&#65039; <span style = "color:red">Note: Checklist for access restriction</span>**

1. List who is in DefaultReader.
2. Create the custom role and assign the specific tables or folders.
3. Add the person to the custom role.
4. Remove them from DefaultReader.
5. Confirm they don’t hold a Contributor, Member, or Admin workspace role.
6. Test as that user. Check that allowed data opens and blocked data fails.
7. Review membership periodically.

**Why it matters**
- The failure mode is silent. Nothing errors, and the user simply sees more than you intended. When you are asked about security in a Fabric interview, “I verify effective permissions by testing as the user, because roles are additive and DefaultReader grants everything” is a strong answer.
- OneLake security is evolving (role types, row and column-level options), so confirm the current behavior in Microsoft’s documentation before applying it to production.


</td>
</table>
