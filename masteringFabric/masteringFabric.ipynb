- You can include a **dataflow** as an **activity** in a **pipeline**. Pipelines are used to **orchestrate** data ingestion and processing **activities**, enabling you to combine dataflows with other kinds of operation in a **single, scheduled process**.
- **Timeout:** Maximum amount of time an activity can run. Default is 12 hours, the minimum is 10 minutes, and the maximum amount of time allowed is 7 days. Format is in D.HH:MM:SS
- **Enable retries:** When enabled, the activity retries on failure based on the retry count, conditions, and interval set within.
    - Retry: Maximum number of retry attempts
    - Retry interval type (Fixed/Increasing Delay): Use Fixed for steady intervals, or Increasing delay to apply a backoff that grows the wait time after each attempt
    - Retry interval (sec): The number of seconds between each retry attempt
    - Retry conditions: Activity will retry only when the defined conditions are met
- **Activity State (Activated/Deactivated):** Deactivate an activity to exclude it from pipeline run and validation.

### **Activator**

- The **monitoring hub** shows you what happened. The **Schedule failures** page emails you when it happens. But neither takes action for you. **Activator** in Microsoft Fabric closes that gap — when a Fabric job event occurs, Activator can automatically notify the right people in the right channel, or kick off a follow-up job without waiting for a human response.
- **When the condition is met, Activator can take these actions:**
    - **Email:** Sends a message with context about the event, including a link to the monitoring hub.
    - **Teams:** Posts a message to an individual, group chat, or channel. This is often more useful than email for on-call teams who need immediate visibility.
    - **Run a Fabric activity:** Triggers a pipeline, notebook, dataflow, spark job, or function.
- **Activator works best for automated coordination:** chaining jobs together, running cleanup actions, or getting the right notification to the right channel — not as a substitute for investigating and fixing the root cause.



<table>
<td>

**The quickest way to connect Activator to a job event is through Real-Time hub, using the Set alert wizard:**

1. In Microsoft Fabric, **select Real-Time** in the left navigation to open Real-Time hub.
2. Select **Fabric events**, then select Job events.
3. Select **Set alert**.
4. In the **Monitor** section, select **Select source events** and complete the wizard — choose the event type (for example, <span style = "color:orange">**Microsoft.Fabric.JobEvents.ItemJobFailed**</span>), then select the workspace and the specific item to monitor.
5. In the **Condition** section, choose when to act — on each matching event, or only when a specific field value is met.
6. In the **Action** section, choose what Activator should do.


</td>
</table>

**&#8505;&#65039; <span style = "color:red">Note: Checklist for access restriction</span>**

1. List who is in DefaultReader.
2. Create the custom role and assign the specific tables or folders.
3. Add the person to the custom role.
4. Remove them from DefaultReader.
5. Confirm they don’t hold a Contributor, Member, or Admin workspace role.
6. Test as that user. Check that allowed data opens and blocked data fails.
7. Review membership periodically.

**Why it matters**
- The failure mode is silent. Nothing errors, and the user simply sees more than you intended. When you are asked about security in a Fabric interview, “I verify effective permissions by testing as the user, because roles are additive and DefaultReader grants everything” is a strong answer.
- OneLake security is evolving (role types, row and column-level options), so confirm the current behavior in Microsoft’s documentation before applying it to production.

# **On-premises Data Gateway**

An on-premises data gateway is a service on a server in your network. It makes outbound connections to Fabric, so you don’t open inbound firewall ports. It works well for modest, scheduled loads, but it has limits.

Limitations, with scenarios

1. Spark notebooks can’t use it

Scenario: A Fabric notebook uses JDBC to read a legacy SQL Server VM and times out. Fabric Spark runs in an isolated managed network, and Dataflows Gen2 and pipelines support on-premises gateways while notebooks don’t. 
microsoft

2. It conflicts with tenant Private Link

Scenario: Security enables Private Link for the tenant and your gateway stops registering. Microsoft’s guidance in one community answer says gateways aren’t supported and fail to register while Private Link is enabled. Verify the current behavior before changing tenant security settings. 
microsoft

3. It’s a server you must run

Scenario: The gateway VM reboots for patching at 5:50 AM and the 6 AM load fails. It is a single point of failure unless clustered, and you maintain updates. For Dataflow Gen2, only the last six released gateway versions are supported, so an unpatched gateway can break refreshes. 
microsoft

4. Throughput is capped by the gateway machine and network

Scenario: A 2 TB historical load that should take hours runs for days. Everything flows through one machine’s CPU, memory, disk, and WAN link. Dataflow Gen2 transformations make this worse for large volumes.

5. Feature gaps

Scenario: You want hourly windowed loads, but tumbling window triggers aren’t yet available in Fabric pipelines, so you build the logic with schedules and watermarks. Feature coverage changes often, and older documentation conflicts with newer behavior, so check the current docs for your connector. 
microsoft

6. Latency and change capture

Scenario: Operations wants claims within minutes. Batch copies through a gateway usually mean scheduled intervals, not continuous change data capture.

7. Credentials and authentication

Scenario: The source needs Windows or Kerberos authentication. Credentials live on the gateway, and delegation and service-account setup is often the slowest part of onboarding.

**Interview Tip:**

- The gateway is outbound-only and easy to set up, but it’s a server to patch and cluster, its throughput is limited by that machine, notebooks can’t use it, and it can conflict with Private Link. I use pipelines or Copy jobs through a clustered gateway to land data in OneLake, run Spark on the staged copy, and use mirroring or Eventstreams when I need low latency.


**Two ways a connection could work**

**Inbound (what the gateway avoids):**

- Fabric (internet)  ──►  your firewall  ──►  SQL Server

    - You would open a port on your firewall, expose the database to the internet, and defend it. Security teams rarely approve this.

**Outbound (how the gateway works):**

- SQL Server ◄──►  Gateway server ──►   Microsoft cloud (Azure relay)   ◄──  Fabric
                (inside network)  outbound only

1. The gateway service starts and dials out to Microsoft’s cloud, usually over HTTPS (port 443) and an Azure messaging relay, and keeps that connection open.
2. When a pipeline or refresh needs on-prem data, Fabric doesn’t call your network. It places a request on the cloud side of that relay.
3. The gateway, which is already connected, picks up the request.
4. It queries the local SQL Server and sends the results back over the same outbound connection.**
